```python
import os,warnings
from dotenv import load_dotenv

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate

from google import genai
from google.genai import types

warnings.filterwarnings("ignore")
```

```python
env = r"D:\stackroute\2_AI-assisted-programming\learning_requirements\bosch\2026\5_advancedPE\code\config.env"
if load_dotenv(env):
    gemini_key = os.getenv("GEMINI_API_KEY")

langsmithkey = os.environ["LANGSMITH_API_KEY"]

# Enable LangSmith Observability / Tracing
os.environ["LANGSMITH_TRACING"] = "true"

# Name under which traces will appear in LangSmith
os.environ["LANGSMITH_PROJECT"] = "Bosch_AdvancedPE"
```

```python
# Create the Gemini client
model = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite",api_key=gemini_key)
```

```python
template = PromptTemplate(template="Give a 10-bullet-point brief on {topic}")

prompt = template.invoke({'topic': 'Model Fine Tuning'})
```

```python
response = model.invoke(prompt)
print(response.text)
```

```python
llm = genai.Client(api_key=gemini_key)

def run(prompt):
    try:
        response = llm.models.generate_content(
            model="gemini-3.1-flash-lite",
            contents=prompt,
            config=types.GenerateContentConfig(
                system_instruction=
                "You are a helpful assistant good in giving information."
            )
        )

        return response

    except Exception as e:
        return "Exception from run().\n" + str(e)
```

```python
response = run("Give a 2-line defition of a Prompt in Generative AI")
```
