# Document Loaders
# load different types of documents for processing

```python
# 1) Text Loader

from langchain_community.document_loaders import TextLoader

file = r"D:\stackroute\2_AI-assisted-programming\learning_requirements\bosch\2026\5_advancedPE\dataset\healthyliving.txt"

# Create the loader
loader = TextLoader(file, encoding="utf-8")

# Read the file
documents = loader.load()

print(documents)
data = documents[0].page_content
```

```python
print(data)
```

```python
# 2) PDF documents

# pip install PyPDF pypdf
from langchain_community.document_loaders import PyPDFLoader
```

```python
file = r"D:\stackroute\2_AI-assisted-programming\learning_requirements\bosch\2026\5_advancedPE\dataset\indian_classical_music.pdf"
loader = PyPDFLoader(file)
data = loader.load()

print(data)
print(f"Total pages = {len(data)}")

# print the individual pages
for i in range(len(data)):
    print(f"Page : {i+1}, Content: \n{data[i].page_content}")
```

```python
# 3) DOCX documents

# %pip install unstructured
# %pip install python-docx
# %pip install docx2txt
from langchain_community.document_loaders import UnstructuredWordDocumentLoader

file = r"D:\stackroute\2_AI-assisted-programming\learning_requirements\bosch\2026\5_advancedPE\dataset\photography.docx"

loader = UnstructuredWordDocumentLoader(file)
documents = loader.load()

data = ''

for doc in documents:
    data+= doc.page_content

print(data)
```

```python
# 4) JSON documents
# python -m pip install jq

from langchain_community.document_loaders import JSONLoader

file = r"D:\stackroute\2_AI-assisted-programming\learning_requirements\bosch\2026\5_advancedPE\dataset\employees.json"

loader = JSONLoader(file_path=file, jq_schema=".[]", text_content=False)

docs = loader.load()

for doc in docs:
    print(doc.page_content)
```
